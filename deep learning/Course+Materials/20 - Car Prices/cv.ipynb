```python
import pandas as pd 
from sklearn.feature_extraction.text import CountVectorizer
```

```python
fd = pd.read_csv("./data/used_cars.csv")
```

```python
cv = CountVectorizer(max_features=4)
```

```python
document=[
    "Hello world,I am Aniket. hello world.",
    "Hello world,I am PUPU"
]
cv.fit(document)
print(cv.get_feature_names_out())
```

```python
out=cv.transform(document)
print(out)
```
