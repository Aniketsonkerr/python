## Measure of central tendencies:

```python
import numpy as np 
from scipy import stats
```

## Setting up the data:

```python
data = np.array([150, 200, 175, 220, 180, 250, 175, 190, 210, 2500])
print(f"Data set of house prices($1000): \n{data}")
```

## Mean: The sum of all data values divided by the total count. It acts as the balance point of the dataset, but it is heavily pulled toward extreme high or low values (outliers) $$\mu = \frac{\sum_{i=1}^{N} x_i}{N} \qquad \text{or} \qquad \bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}$$

```python
## mean is heavily influenced by outliers 
mean_of_prices = np.mean(data)
print(f"The mean price from the given data is {mean_of_prices}")
```

## Median 1.Sort all $n$ data points from smallest to largest. 2.Calculate the positional index $k$: 1.Odd $n$: The exact middle value at position $\frac{n + 1}{2} $.2.Even $n$: The average of the two middle values at positions $\frac{n}{2}$ and $\left(\frac{n}{2} + 1\right)$:$$\text{Median} = \frac{x_{\left(\frac{n}{2}\right)} + x_{\left(\frac{n}{2} + 1\right)}}{2}$$

```python
median_of_prices = np.median(data)
print(f"the median price from the given data is {median_of_prices}")
```
