# Statistics:

## 1. Descriptive stats:

### 1.1 mean:

```python
#lets take an array of prices of houses :
array = np.random.randint(400,600, size=10)
print(array)
```

```python
npArray = np.array(array);
npArray
```

```python
mean_of_house_prices = np.mean(npArray)
print(f"The mean for house prices is :\n {mean_of_house_prices}")
```

### 1.2 median

```python
#lets take the median of the same data :
# it's middle most value of the data 
median = np.median(npArray)
print(f"the median price of the houses:\n {median}")
```
