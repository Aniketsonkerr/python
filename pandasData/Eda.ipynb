```python
import pandas as pd 
import numpy as np
from sklearn.impute import SimpleImputer
```

## 1.Data extraction by using pandas:

```python
data = pd.read_csv("googleplaystore.csv")
data.shape #no. of rows and columns
```

## 2.Find number of NULL datas:

```python
data.isna().sum()
```

## 3. Removing NULL values from a column of datas:

```python
imputer = SimpleImputer(missing_values=np.nan,strategy="most_frequent")
imputer.fit(data.iloc[:,3:4].values)
data.iloc[:,3:4]=imputer.transform(data.iloc[:,3:4].values)
data.isna().sum()
```

## 4. Removing NUll values from all the DATA and replacing it by mean:

```python
imputer = SimpleImputer(missing_values=np.nan,strategy="most_frequent")
imputer.fit(data.iloc[:,:].values)
data.iloc[:,:]=imputer.transform(data.iloc[:,:].values)
data
```
