```python
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
housing = fetch_california_housing()
```

```python
print(housing.DESCR)
```

```python
dataset = pd.DataFrame(housing.data,columns=housing.feature_names)
dataset.head()
```

```python
dataset.info()
```

```python
dataset.isnull().sum()
```

```python
dataset["Price"]= housing.target
```

```python
dataset.head()
```

```python
dataset.describe()
```

```python
dataset.corr() #to see if datas are co-related range is (-1 to 1) where 1 and-1 is veryyyy high 
```

```python
sns.pairplot(dataset)
```

```python
fig, ax = plt.subplots(figsize=(15,15))
sns.boxplot(data=dataset,ax=ax)
plt.savefig("boxplot.jpg")
```

```python
# now lets normalise the data as the population will bias the prediction and before normalising the data let's split the data for training and test 
from sklearn.model_selection import train_test_split
```

```python
x_features = dataset.iloc[:,:-1]
y_target = dataset.iloc[:,-1]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_features,y_target,train_size=0.8,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scalar= StandardScaler()
```

```python
x_train_norm = scalar.fit_transform(x_train)
```

```python
fig, ax = plt.subplots(figsize=(15,15))
sns.boxplot(data=x_train_norm,ax=ax)
plt.savefig("boxplot_norm_data.jpg")
```

```python
x_test_transformed = scalar.transform(x_test)
```

```python
fig, ax = plt.subplots(figsize=(15,15))
sns.boxplot(data=x_test_transformed,ax=ax)
plt.savefig("boxplot_x_test_trans.jpg")
```

```python
#model training
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression.fit(x_train_norm,y_train)
```

```python
reg_pred = regression.predict(x_test_transformed)
```

```python
print(reg_pred)
```

```python
residual = y_test-reg_pred
```

```python
residual
```

```python
sns.displot(residual,kind="kde")
```
